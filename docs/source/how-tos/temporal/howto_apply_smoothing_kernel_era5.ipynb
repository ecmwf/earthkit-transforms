# Apply a Smoothing Kernel

This notebook demonstrates how to apply a smoothing kernel to some sample 2m temperature from ERA5 data using earthkit-data and earthkit-transforms.
Smoothing kernels are useful for filtering high-frequency noise while preserving longer-term variability in a time series.

```python
import numpy as np
from earthkit import data as ekd
from earthkit import transforms as ekt
from earthkit.transforms._tools import earthkit_remote_test_data_file

remote_era5_file = earthkit_remote_test_data_file("era5-Europe-sfc-2m-temperature-3deg-2015-2017.grib")
era5_data = ekd.from_source("url", remote_era5_file)
ds = era5_data.to_xarray()

# ~7-day normalised Hann window (29 six-hourly steps)
window = np.hanning(29)
window = window / window.sum()

smoothed = ekt.temporal.convolve(ds, window, remove_partial_periods=True)
smoothed
```
